## ChipWhisperer Hardware Test

The following will connect to a ChipWhisperer. Just plug the Chipwhisperer-Husky main board in. If the ChipWhisperer communication is up, everything else should "just work" when talking to various targets.

This should result in some text that says ``INFO: Found ChipWhisperer😍``

```python
import chipwhisperer as cw
scope = cw.scope()
print("INFO: Found ChipWhisperer😍")
```

We aren't going to use the hardware, so should disconnect from it so it is dropped by this Python kernel.
