```python
import numpy as np

list1=[]
list2=[]
i=0
for x in range(0,3):
    items = []
    for _ in range(0,10):
        i += 1
        items.append(i)
    list1.append(items)
i += 33
for x in range(0,3):
    items = []
    for _ in range(0,10):
        i += 2
        items.append(i)
    list2.append(items)
    
print(type(items))
print(items)
a = np.asarray(list1)
print(type(a))
print(a)
list1_avg = a.mean(axis=0)
print("list1_avg ",list1_avg)

b = np.asarray(list2)
print(type(b))
print(b)
list2_avg = b.mean(axis=0)
print("list2_avg ",list2_avg)

print( "abs(list1_avg - list2_avg) " , abs(list1_avg - list2_avg))

print( "np.max(abs(list1_avg - list2_avg)) " , np.max(abs(list1_avg - list2_avg)))
```
