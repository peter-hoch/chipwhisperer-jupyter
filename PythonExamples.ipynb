# Python examples

## sliceing

```python
s = "012345"
print("s= ", s)
print("s[0:5]", s[0:5])
print("s[2:5]", s[2:5])
print("s[:3]", s[:3])
print("s[3:]", s[3:])
print("s[-2]", s[-2])
print("s[-2:]", s[-2:])
print("s[:-2]", s[:-2])
print("s[::2]", s[::2])
print("s[::-1]", s[::-1])
print("s[1:5:2]", s[1:5:2])
print("s[:,0]", s[:0])
```
